Build a small nn.Module with one nn.Linear(2, 4) layer followed by nn.ReLU() . Pass the tensor [1.0, -1.0] through it
and print the output. In 1-2 lines, explain what ReLU did to any negative values.


```python
# import the proper libraries for this we are using torch or the PyTorch
import torch
import torch.nn as nn
```

```python
class small(nn.Module):
    def __init__(self):
        super(small, self).__init__()
        self.hidden = nn.Linear(2, 4)  # input
        self.output = nn.Linear(4, 1)  #

    def forward(self, x):
        x = torch.relu(self.hidden(x))  # with ReLU activation
        x = self.output(x)
        return x
```

```python
# Initilizing the model
model = small()
```

```python
x = torch.tensor([[1.0, -1.0]])
y = model(x)
print("utput with torch.nn:", y)
```

Load gpt2 and generate two outputs for the same prompt, "Once upon a time": one with do_sample=False and one with
do_sample=True . Print both and note one difference between them.


```python
# This code is refer by huggingface another document
# importing the transormer library to generate the code.
from transformers import AutoTokenizer, AutoModelForCausalLM, WatermarkDetector, WatermarkingConfig

model = AutoModelForCausalLM.from_pretrained("gpt2")
tok = AutoTokenizer.from_pretrained("gpt2")
tok.pad_token_id = tok.eos_token_id
tok.padding_side = "left"

inputs = tok(["Once upon a time"], padding=True, return_tensors="pt")
input_len = inputs["input_ids"].shape[-1]

watermarking_config = WatermarkingConfig(bias=2.5, seeding_scheme="selfhash")
out = model.generate(**inputs, watermarking_config=watermarking_config, do_sample=True, max_length=20) # I use both true or false

detector = WatermarkDetector(model_config=model.config, device="cpu", watermarking_config=watermarking_config) # defining where we are running
detection_out = detector(out, return_dict=True)
detection_out.prediction

# I know the fewer concept but now i dont know the watermarking in the huggingface code
```
